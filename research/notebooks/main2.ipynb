# Uncertainty benchmark against exact (reference) values

Every labelled system is fitted on **all** of its finite-basis rows with each family
(`exponential`, `sqrt_exponential`, `power_law`) and each B mode (`free`; `energy` = B imposed
from the Energy fit, B/2 for singular operators). The resulting asymptote $C \pm \sigma$ is
compared with the reference value $C_{\rm ref} \pm \sigma_{\rm ref}$.

**Scored systems**: the 11 files in `large-dataset/` (reference = last row, its error from
`large-dataset-err/`) and the 3 files in `new-dataset/` (reference = second-to-last row, error =
last row). `muon/` and `small-dataset/` have no reference values and are not scored.

**Per-case table** (one row per system × observable × b_mode × family)

| column | meaning |
|---|---|
| `C_fit`, `sigma` | extrapolated limit and its uncertainty, in original units |
| `C_ref`, `sigma_ref` | reference value and its error bar |
| `dC` | `C_fit - C_ref` (signed error) |
| `rho` | `abs(dC) / sigma`: how many error bars the true error is |
| `rho_eff` | `max(abs(dC) - sigma_ref, 0) / sigma`: the same, crediting the reference's own error bar; `rho_eff <= 1` ⇔ the intervals `C_fit ± sigma` and `C_ref ± sigma_ref` overlap |
| `ref_limited` | `abs(dC) <= sigma_ref`: the fit is already closer than the reference can resolve, so the case cannot test `sigma` |
| `verdict` | `too narrow` (rho_eff > 3) · `too wide` (sigma > 10·max(abs(dC), sigma_ref)) · `sensible` (neither) · `inf` (no finite sigma) · `ref-limited` |

**Summary metrics** (per family × b_mode, then per system) are computed on the testable,
finite-σ cases only; the excluded counts are reported alongside.

```python
import contextlib, io, warnings
from scipy.stats import spearmanr

DATA = '../data'

# name, file, reference/error layout, rows dropped from the end, drop SO1/SO2
SYSTEMS = [
    dict(name='Be_1De',     path='large-dataset/be_1De.xls',      err='large-dataset-err/be_1De_err.xls',     cut=1, drop_so=True),
    dict(name='Be_3De',     path='large-dataset/be_3De.xls',      err='large-dataset-err/be_3De_err.xls',     cut=1, drop_so=False),
    dict(name='Be_3Pe',     path='large-dataset/be_3Pe.xls',      err='large-dataset-err/be_3Pe_err.xls',     cut=1, drop_so=False),
    dict(name='Li_2Se',     path='large-dataset/Li_2Se.csv',      err='large-dataset-err/li_2se_err.csv',     cut=1, drop_so=True),
    dict(name='Be_1Se',     path='large-dataset/Be_1Se.csv',      err='large-dataset-err/be_1se_err.csv',     cut=1, drop_so=True),
    dict(name='Be_3Po',     path='large-dataset/Be_3Po.csv',      err='large-dataset-err/be_3po_err.csv',     cut=1, drop_so=True),
    dict(name='Be_1Po',     path='large-dataset/Be_1Po.csv',      err='large-dataset-err/be_1po_err.csv',     cut=1, drop_so=True),
    dict(name='Li_2Po',     path='large-dataset/Li_2Po.csv',      err='large-dataset-err/li_2po_err.csv',     cut=1, drop_so=True),
    dict(name='Be_3Se',     path='large-dataset/Be_3Se.csv',      err='large-dataset-err/be_3se_err.csv',     cut=1, drop_so=True),
    dict(name='Be_1Se_1Po', path='large-dataset/Be_1Se_1Po.xls',  err='large-dataset-err/be_1se_1po_err.csv', cut=3, drop_so=False),
    dict(name='Li_2Se_2Po', path='large-dataset/Li_2Se_2Po.xls',  err='large-dataset-err/li_2se_2po_err.csv', cut=3, drop_so=False),
    dict(name='Be_3Po_new', path='new-dataset/be_3po_test.csv',    err=None, cut=2, drop_so=False),
    dict(name='Be_3Se_new', path='new-dataset/be_3se_test.csv',    err=None, cut=2, drop_so=False),
    dict(name='Be_3Se_01_new', path='new-dataset/be_3se_01_test.csv', err=None, cut=2, drop_so=False),
]


def load_system(spec):
    """Return (fit rows, reference row, reference-error row), all with a 'basis size' column."""
    df = upload_df(f"{DATA}/{spec['path']}").rename(columns={'Basis Size': 'basis size'})
    if spec['err'] is None:                       # new-dataset: [..., reference, error]
        ref, err = df.iloc[-2], df.iloc[-1]
    else:                                         # large-dataset: [..., reference] + separate error file
        ref = df.iloc[-1]
        err = pd.read_csv(f"{DATA}/{spec['err']}").iloc[0]
    if spec['drop_so']:
        df = df.drop(columns=['SO1', 'SO2'], errors='ignore')
    return df.iloc[:-spec['cut']].reset_index(drop=True), ref, err


systems = {s['name']: load_system(s) for s in SYSTEMS}
pd.DataFrame([
    dict(system=k, n_pts=len(d), basis_min=int(d['basis size'].min()), basis_max=int(d['basis size'].max()),
         ref_basis=int(r['basis size']), n_observables=len(d.columns) - 1)
    for k, (d, r, e) in systems.items()
])
```

```python
FAMILIES = ['exponential', 'sqrt_exponential', 'power_law']


def run_case(df, obs, b_mode):
    """Fit one observable with all three families; return {family: (C, sigma, floor_bound)}, original units."""
    with contextlib.redirect_stdout(io.StringIO()), warnings.catch_warnings():
        warnings.simplefilter('ignore')
        solver = extraplus(df, 'basis size', obs, use_energy_b=(b_mode == 'energy'))
        res = solver.fit_linearized(models=FAMILIES)
    return {m: (solver.y_min + solver.y_range * r['C'], solver.y_range * r['sigma_C'],
                r['sigma_model'] <= r['sigma_num'])
            for m, r in res.items()}


def verdict(row):
    if not np.isfinite(row['sigma']):
        return 'inf'
    if row['ref_limited']:
        return 'ref-limited'
    if row['rho_eff'] > 3:
        return 'too narrow'
    if row['sigma'] > 10 * max(abs(row['dC']), row['sigma_ref']):
        return 'too wide'
    return 'sensible'


rows = []
for name, (df, ref, err) in systems.items():
    has_energy = 'Energy' in df.columns
    for obs in df.columns.drop('basis size'):
        y = df[obs].to_numpy(dtype=float)
        if not np.isfinite(y).all() or np.ptp(y) == 0 or obs not in ref.index:
            continue
        modes = ['free'] + (['energy'] if has_energy and obs != 'Energy' else [])
        for b_mode in modes:
            for fam, (C, sig, floor) in run_case(df, obs, b_mode).items():
                rows.append(dict(system=name, observable=obs, b_mode=b_mode, family=fam, n_pts=len(y),
                                 C_fit=C, sigma=sig, floor_bound=floor, C_ref=float(ref[obs]),
                                 sigma_ref=float(err[obs]) if obs in err.index else np.nan))

bench = pd.DataFrame(rows)
bench['dC'] = bench['C_fit'] - bench['C_ref']
bench['rho'] = bench['dC'].abs() / bench['sigma']
bench['rho_eff'] = (bench['dC'].abs() - bench['sigma_ref'].fillna(0)).clip(lower=0) / bench['sigma']
bench['ref_limited'] = bench['dC'].abs() <= bench['sigma_ref']          # NaN sigma_ref -> False
bench['verdict'] = bench.apply(verdict, axis=1)
print(f"{len(bench)} cases: {bench['system'].nunique()} systems, "
      f"{bench.groupby(['system', 'observable']).ngroups} system-observable pairs, "
      f"{bench['sigma_ref'].isna().sum()} cases without a reference error")
```

### Per-case table
Numbers in scientific notation; filter `bench` as needed, e.g. `bench.query("system == 'Be_1De'")`.

```python
CASE_COLS = ['system', 'observable', 'b_mode', 'family', 'n_pts', 'C_fit', 'sigma', 'C_ref', 'sigma_ref',
             'dC', 'rho', 'rho_eff', 'verdict']
with pd.option_context('display.max_rows', 40, 'display.float_format', '{:.4e}'.format):
    display(bench[CASE_COLS])
```

```python
def summarize(g):
    """Sensibility metrics for one group of cases (testable = finite sigma and not reference-limited)."""
    fin = g[np.isfinite(g['sigma'])]
    t = fin[~fin['ref_limited']]
    r = t['rho_eff']
    return pd.Series({
        'cases':           len(g),
        'sigma_inf_%':     100 * (~np.isfinite(g['sigma'])).mean(),
        'ref_limited':     int(fin['ref_limited'].sum()),
        'floor_bound':     int(g['floor_bound'].sum()),
        'testable':        len(t),
        'covered_1σ_%':    100 * (r <= 1).mean(),
        'covered_2σ_%':    100 * (r <= 2).mean(),
        'covered_3σ_%':    100 * (r <= 3).mean(),
        'median_rho_eff':  r.median(),
        'too_narrow_%':    100 * (r > 3).mean(),
        'sensible_%':      100 * (t['verdict'] == 'sensible').mean(),
        'too_wide_%':      100 * (t['verdict'] == 'too wide').mean(),
        'max_rho_eff':     r.max(),
    })


fmt = {c: '{:.1f}' for c in ['sigma_inf_%', 'covered_1σ_%', 'covered_2σ_%', 'covered_3σ_%',
                             'too_narrow_%', 'sensible_%', 'too_wide_%']}
fmt.update({'median_rho_eff': '{:.3g}', 'max_rho_eff': '{:.3g}',
            'cases': '{:.0f}', 'ref_limited': '{:.0f}', 'floor_bound': '{:.0f}', 'testable': '{:.0f}'})

overall = pd.concat([bench.groupby(['family', 'b_mode']).apply(summarize, include_groups=False),
                     summarize(bench).to_frame(('ALL', 'ALL')).T])
display(overall.style.format(fmt).set_caption('Uncertainty sensibility by family and B mode'))
```

```python
by_system = bench.groupby(['system', 'family']).apply(summarize, include_groups=False)
display(by_system[['cases', 'sigma_inf_%', 'testable', 'covered_1σ_%', 'covered_3σ_%', 'median_rho_eff',
                   'too_narrow_%', 'too_wide_%']]
        .style.format(fmt).set_caption('Uncertainty sensibility by system and family (both B modes)'))
```

```python
# Does sigma rank the three families by accuracy?  Within one (system, observable, b_mode) the
# observable's scale is fixed, so this tests sigma itself rather than the magnitude of the property.
# Only groups where the ranking is decidable: 3 finite, distinct sigmas and no reference-limited member.
def family_ranking(g):
    if (len(g) < 3 or not np.isfinite(g['sigma']).all() or g['sigma'].nunique() < 3
            or g['ref_limited'].any() or g['sigma_ref'].isna().any()):
        return None
    err = g['dC'].abs()
    return pd.Series({'smallest_sigma_is_most_accurate': g['sigma'].idxmin() == err.idxmin(),
                      'largest_sigma_is_least_accurate': g['sigma'].idxmax() == err.idxmax(),
                      'spearman(σ,|dC|)':                spearmanr(g['sigma'], err).statistic})


with warnings.catch_warnings():                 # spearman of a constant triple is NaN; skip it
    warnings.simplefilter('ignore')
    ranking = (bench.groupby(['system', 'observable', 'b_mode'])
               .apply(family_ranking, include_groups=False).dropna(how='all').astype(float))
n_groups = bench.groupby(['system', 'observable', 'b_mode']).ngroups
rank_summary = ranking.groupby(level='b_mode').mean().assign(groups=ranking.groupby(level='b_mode').size())
print(f"{len(ranking)} of {n_groups} (system, observable, b_mode) groups are decidable")
display(rank_summary.style.format('{:.2f}', subset=rank_summary.columns[:-1])
        .set_caption('Family ranking by σ vs by true error (chance: 0.33, 0.33, 0.00)'))
```

## Benchmark v2: ρ matrix per system

`benchmark_all(systems)` takes, for every system, the finite-basis rows plus the **exact value** (`inf_df`) and its **uncertainty** (`err_df`). It fits every observable with the three families and returns one tidy table with `rho` for each (system, observable, B mode, family). `rho_matrix` then lays the result out as one block per system.

$$\rho=\frac{|C_{\rm fit}-C_{\rm exact}|}{\sigma},\qquad \rho_{\rm eff}=\frac{\max(|C_{\rm fit}-C_{\rm exact}|-\sigma_{\rm exact},\,0)}{\sigma}$$

| ρ | reading |
|---|---|
| ≤ 1 | true error inside the quoted σ (covered) |
| 1 – 3 | σ too small, but not wildly |
| > 3 | σ too narrow |
| ≪ 0.1 | σ much wider than the real error (uninformative) |
| grey | $|C_{\rm fit}-C_{\rm exact}|\le\sigma_{\rm exact}$: the fit is already closer than the exact value can resolve, so this cell cannot test σ |

`inf_df` and `err_df` may each be a one-row DataFrame or a Series. The cells reuse the `systems` dict from the first benchmark (14 systems: `(fit rows, exact row, error row)`).

```python
import contextlib, io, warnings
import numpy as np
import pandas as pd
from IPython.display import display, HTML

FAMILIES  = ['exponential', 'sqrt_exponential', 'power_law']
FAM_SHORT = {'exponential': 'exp', 'sqrt_exponential': 'sqrt-exp', 'power_law': 'power'}
X_COL     = 'basis size'


def _as_row(obj):
    """inf_df / err_df can be a one-row DataFrame or a Series; return a Series."""
    return obj.iloc[-1] if isinstance(obj, pd.DataFrame) else obj


def benchmark_system(init_df, inf_df, err_df, name='system', b_modes=('free', 'energy'), x_col=X_COL):
    """Fit every observable of one system with all families and compare with the exact values.

    init_df : finite-basis rows (x_col + observables)
    inf_df  : exact (reference) values, one row / Series
    err_df  : uncertainties of the exact values, one row / Series
    Returns a long DataFrame, one row per (observable, b_mode, family), original units.
    """
    inf_row, err_row = _as_row(inf_df), _as_row(err_df)
    has_energy = 'Energy' in init_df.columns
    rows = []
    for obs in init_df.columns.drop(x_col):
        y = init_df[obs].to_numpy(dtype=float)
        if obs not in inf_row.index or not np.isfinite(y).all() or np.ptp(y) == 0:
            continue
        c_exact = float(inf_row[obs])
        if not np.isfinite(c_exact):
            continue
        s_exact = float(err_row[obs]) if obs in err_row.index else np.nan
        for b_mode in b_modes:
            if b_mode == 'energy' and (not has_energy or obs == 'Energy'):
                continue
            base = dict(system=name, observable=obs, b_mode=b_mode, n_pts=len(y),
                        C_exact=c_exact, sigma_exact=s_exact)
            try:
                with contextlib.redirect_stdout(io.StringIO()), warnings.catch_warnings():
                    warnings.simplefilter('ignore')
                    solver = extraplus(init_df, x_col, obs, use_energy_b=(b_mode == 'energy'))
                    res = solver.fit_linearized(models=FAMILIES)
                for fam in FAMILIES:
                    r = res[fam]
                    rows.append(dict(base, family=fam,
                                     C_fit=solver.y_min + solver.y_range * r['C'],
                                     sigma=solver.y_range * r['sigma_C']))
            except Exception as exc:                      # keep the failure visible, do not abort the run
                print(f'[fit failed] {name} / {obs} / {b_mode}: {type(exc).__name__}: {exc}')
                rows += [dict(base, family=fam, C_fit=np.nan, sigma=np.nan) for fam in FAMILIES]
    return pd.DataFrame(rows)


def benchmark_all(systems, b_modes=('free', 'energy')):
    """systems: {name: (init_df, inf_df, err_df)}  ->  long DataFrame with rho for every case."""
    parts = [benchmark_system(init, inf, err, name=name, b_modes=b_modes)
             for name, (init, inf, err) in systems.items()]
    out = pd.concat(parts, ignore_index=True)
    out['dC']          = out['C_fit'] - out['C_exact']
    out['rho']         = out['dC'].abs() / out['sigma']
    out['rho_eff']     = (out['dC'].abs() - out['sigma_exact'].fillna(0)).clip(lower=0) / out['sigma']
    out['ref_limited'] = out['dC'].abs() <= out['sigma_exact']          # NaN sigma_exact -> False
    return out
```

```python
def _best_family(g, col, absolute=False):
    """Short name of the family with the smallest `col` (abs value if requested)."""
    v = g.set_index('family')[col]
    v = v.abs() if absolute else v
    v = v[np.isfinite(v)]
    return FAM_SHORT[v.idxmin()] if len(v) else '–'


def rho_matrix(bench, system, b_mode='free'):
    """One block: rows = observables, columns = rho of each family (+ exact value, best family)."""
    g = bench[(bench['system'] == system) & (bench['b_mode'] == b_mode)]
    if g.empty:
        return None
    order = list(dict.fromkeys(g['observable']))
    rho = g.pivot(index='observable', columns='family', values='rho').reindex(index=order, columns=FAMILIES)
    lim = (g.pivot(index='observable', columns='family', values='ref_limited')
            .reindex(index=order, columns=FAMILIES).fillna(False).astype(bool))
    first = g.drop_duplicates('observable').set_index('observable').reindex(order)
    by_obs = {o: d for o, d in g.groupby('observable')}

    rho_cols = [f'ρ {FAM_SHORT[f]}' for f in FAMILIES]
    out = pd.DataFrame({'C_exact': first['C_exact'], 'σ_exact': first['sigma_exact']})
    for f, c in zip(FAMILIES, rho_cols):
        out[c] = rho[f]
    out['smallest σ']    = [_best_family(by_obs[o], 'sigma') for o in order]
    out['smallest |ΔC|'] = [_best_family(by_obs[o], 'dC', absolute=True) for o in order]
    out.index.name = 'observable'
    lim.columns = rho_cols

    def colour(sub):
        css = pd.DataFrame('', index=sub.index, columns=sub.columns)
        for c in sub.columns:
            for o in sub.index:
                v = sub.at[o, c]
                if lim.at[o, c]:
                    css.at[o, c] = 'background-color:#e0e0e0;color:#777'
                elif not np.isfinite(v):
                    css.at[o, c] = 'background-color:#f0f0f0'
                elif v <= 1:
                    css.at[o, c] = 'background-color:#c8e6c9'
                elif v <= 3:
                    css.at[o, c] = 'background-color:#ffe0a3'
                else:
                    css.at[o, c] = 'background-color:#f4b6b6'
        return css

    return (out.style.apply(colour, axis=None, subset=rho_cols)
               .format({'C_exact': '{:.10g}', 'σ_exact': '{:.2e}', **{c: '{:.3g}' for c in rho_cols}},
                       na_rep='–'))


def show_rho_matrices(bench, b_mode='free', only=None):
    print('green ρ ≤ 1 · amber 1–3 · red > 3 · grey = |ΔC| ≤ σ_exact (not testable)')
    for name in (only or list(dict.fromkeys(bench['system']))):
        m = rho_matrix(bench, name, b_mode)
        if m is not None:
            display(HTML(f'<h4 style="margin:14px 0 2px">{name} &nbsp;·&nbsp; B mode: {b_mode}</h4>'))
            display(m)
```

### Conclusion: how sensible are the uncertainties (v2)
Computed from `bench2`, so it follows any re-run. *Testable* = finite σ and not grey in the matrices. Coverage uses $\rho_{\rm eff}$ (credits the exact value's own error bar); the median and 90th-percentile columns use plain ρ. The 90th percentile of ρ is the factor by which σ would have to be multiplied to cover 90 % of cases. *Too wide* = σ > 10 × max(|ΔC|, σ_exact).

```python
def sensibility_metrics(g):
    fin = g[np.isfinite(g['sigma']) & np.isfinite(g['rho'])]
    t = fin[~fin['ref_limited']]
    too_wide = t['sigma'] > 10 * np.maximum(t['dC'].abs(), t['sigma_exact'].fillna(0))
    return pd.Series({
        'cases':          len(g),
        'σ not finite %': 100 * (1 - len(fin) / len(g)),
        'not testable':   int(fin['ref_limited'].sum()),
        'testable':       len(t),
        'covered ≤1σ %':  100 * (t['rho_eff'] <= 1).mean(),
        'covered ≤3σ %':  100 * (t['rho_eff'] <= 3).mean(),
        'median ρ':       t['rho'].median(),
        '90th pct ρ':     t['rho'].quantile(0.90),
        'too narrow %':   100 * (t['rho_eff'] > 3).mean(),
        'too wide %':     100 * too_wide.mean(),
    })


metrics = pd.concat([bench2.groupby(['family', 'b_mode']).apply(sensibility_metrics, include_groups=False),
                     sensibility_metrics(bench2).to_frame(('ALL', 'ALL')).T])
fmt2 = {c: '{:.1f}' for c in ['σ not finite %', 'covered ≤1σ %', 'covered ≤3σ %', 'too narrow %', 'too wide %']}
fmt2.update({'median ρ': '{:.3g}', '90th pct ρ': '{:.3g}',
             'cases': '{:.0f}', 'not testable': '{:.0f}', 'testable': '{:.0f}'})
display(metrics.style.format(fmt2)
        .background_gradient(subset=['covered ≤1σ %', 'covered ≤3σ %'], cmap='RdYlGn', vmin=0, vmax=100)
        .background_gradient(subset=['too narrow %'], cmap='RdYlGn_r', vmin=0, vmax=100)
        .set_caption('Uncertainty sensibility by family × B mode (v2)'))
```
